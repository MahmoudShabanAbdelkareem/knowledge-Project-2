```python
def run_project1():
    print("A | B | C | D | E | F | Status")
    print("-" * 35)
    options = [True, False]
    final_solution = None
    for a in options:
        for b in options:
            for c in options:
                for d in options:
                    for e in options:
                        for f in options:
                            all_vals = [a, b, c, d, e, f]
                            liars_count = all_vals.count(False)
                            check_a = (a == (not b))
                            check_b = (b == (c and d))
                            check_c = (c == (not e if a else True))
                            check_d = (d == (a == b))
                            check_e = (e == (liars_count >= 2))
                            check_f = (f == (d and not e))
                            valid_group = (True in all_vals) and (False in all_vals)
                            is_correct = all([check_a, check_b, check_c, check_d, check_e, check_f, valid_group])
                            row = " | ".join(['T' if v else 'F' for v in all_vals])
                            if is_correct:
                                print(f"{row} | correct")
                                final_solution = all_vals
                            else:
                                print(f"{row} | X")
    print("-" * 35)
    if final_solution:
        print("FINAL ESCAPE CONDITION :")
        names = ['A', 'B', 'C', 'D', 'E', 'F']
        for i in range(6):
            status = "Truth-teller" if final_solution[i] else "Liar"
            print(f"Traveler {names[i]}: {status}")

run_project1()
```

```python
def english_to_predicate(sentence):
    words = sentence.split()

    if words[0] == "All" or words[0] == "every":
        subject = words[1]
        adjective = words[3]
        return "Ax(" + subject + "(x) -> " + adjective + "(x))"

    return "Format not supported"

def predicate_to_english(expression):
    if "Ax" in expression:
        step1 = expression.replace("Ax(", "")
        step2 = step1.replace("))", "")
        parts = step2.split("(x) -> ")

        # Added a check to prevent IndexError if the '->' pattern is missing
        if len(parts) < 2:
            return "Format not supported: Could not parse subject and adjective. Check for '->' pattern."

        subject = parts[0]
        adjective = parts[1].replace("(x", "")
        return "All " + subject + "s are " + adjective + "."

    return "Format not supported"

print("Type 'e' to convert English to Logic")
print("Type 'p' to convert Logic to English")
choice = input("Your choice: ")

if choice == 'e':
    eng_text = input("English: ")
    print("Logic: " + english_to_predicate(eng_text))
if choice == 'p':
    logic_text = input("Logic: ")
    print("English: " + predicate_to_english(logic_text))
```
